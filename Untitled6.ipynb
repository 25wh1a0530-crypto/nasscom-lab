```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

```

```python
import os
print(os.listdir())
```

```python
df = pd.read_excel('seasonal_agriculture_performance_dataset (3).xlsx')
```

```python
print("Shape:", df.shape)
```

```python
# Sanity checks on ranges
print("Negative Yield:", (df['Yield_Tonnes_Ha'] < 0).sum())
print("Soil_pH out of 0-14 range:", ((df['Soil_pH'] < 0) | (df['Soil_pH'] > 14)).sum())
print("Humidity above 100%:", (df['Humidity_pct'] > 100).sum())
print("Negative Farm_Area:", (df['Farm_Area_Hectares'] <= 0).sum())
```

```python

missing = df.isnull().sum()
missing = missing[missing > 0]
dupes = df.duplicated(subset='Farm_ID').sum()
bad_ph = ((df['Soil_pH'] < 0) | (df['Soil_pH'] > 14)).sum()
bad_humidity = (df['Humidity_pct'] > 100).sum()
bad_yield = (df['Yield_Tonnes_Ha'] < 0).sum()

print("Columns with missing values:\n", missing if len(missing) > 0 else "None")
print("Duplicate Farm_IDs:", dupes)
print("Invalid Soil_pH rows:", bad_ph)
print("Invalid Humidity rows:", bad_humidity)
print("Negative Yield rows:", bad_yield)
```

```python
initial_rows = len(df)
df = df.drop_duplicates(subset='Farm_ID')
df = df[(df['Soil_pH'] >= 0) & (df['Soil_pH'] <= 14)]
df = df[df['Humidity_pct'] <= 100]
df = df[df['Yield_Tonnes_Ha'] >= 0]
df = df.dropna()
final_rows = len(df)

print(f"Rows before cleaning: {initial_rows}")
print(f"Rows after cleaning: {final_rows}")
print(f"Rows removed: {initial_rows - final_rows}")
```

```python
season_summary = df.groupby('Season')[['Yield_Tonnes_Ha', 'Profit_INR',
                                         'Water_Used_m3', 'Disease_Pest_Risk_pct']].mean()
print(season_summary)
print()

best_profit_season = season_summary['Profit_INR'].idxmax()
best_yield_season = season_summary['Yield_Tonnes_Ha'].idxmax()
most_water_season = season_summary['Water_Used_m3'].idxmax()
most_risk_season = season_summary['Disease_Pest_Risk_pct'].idxmax()

print(f"→ {best_profit_season} shows the highest average profit ({season_summary['Profit_INR'].max():,.0f} INR/farm)")
print(f"→ {best_yield_season} shows the highest average yield ({season_summary['Yield_Tonnes_Ha'].max():.2f} t/ha)")
print(f"→ {most_water_season} uses the most water on average ({season_summary['Water_Used_m3'].max():,.0f} m3)")
print(f"→ {most_risk_season} carries the highest disease/pest risk ({season_summary['Disease_Pest_Risk_pct'].max():.1f}%)")
```

```python
numeric_cols = df.select_dtypes(include=[np.number]).columns
corr_matrix = df[numeric_cols].corr()
yield_corr = corr_matrix['Yield_Tonnes_Ha'].sort_values(ascending=False)
print(yield_corr)

top_driver = yield_corr.index[1]  # index 0 is Yield itself
top_value = yield_corr.iloc[1]
print(f"\n→ {top_driver} shows the strongest relationship with Yield_Tonnes_Ha (r = {top_value:.2f})")
```

```python
def test_seasonal_difference(column):
    groups = [df[df['Season'] == s][column] for s in df['Season'].unique()]
    f_stat, p_value = stats.f_oneway(*groups)
    significant = "statistically significant" if p_value < 0.05 else "not statistically significant"
    print(f"{column}: F={f_stat:.3f}, p={p_value:.5f} → seasonal difference is {significant}")
    return f_stat, p_value

print("Chosen technique: One-way ANOVA — appropriate because we're comparing means")
print("across 3 independent season groups (Kharif/Rabi/Zaid) for a numeric outcome.\n")

test_seasonal_difference('Profit_INR')
test_seasonal_difference('Yield_Tonnes_Ha')
test_seasonal_difference('Water_Used_m3')
test_seasonal_difference('Disease_Pest_Risk_pct')
```

```python
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

sns.boxplot(x='Season', y='Profit_INR', data=df, ax=axes[0,0])
axes[0,0].set_title('Profit by Season')

sns.barplot(x='Season', y='Water_Used_m3', data=df, estimator=np.mean, ax=axes[0,1])
axes[0,1].set_title('Average Water Used by Season')

sns.boxplot(x='Season', y='Disease_Pest_Risk_pct', data=df, ax=axes[1,0])
axes[1,0].set_title('Disease/Pest Risk by Season')

sns.boxplot(x='Season', y='Yield_Tonnes_Ha', data=df, ax=axes[1,1])
axes[1,1].set_title('Yield by Season')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,8))
sns.heatmap(corr_matrix, annot=False, cmap='coolwarm', center=0)
plt.title('Correlation Heatmap')
plt.show()
```

```python
state_season_profit = df.groupby(['State', 'Season'])['Profit_INR'].mean().unstack()
print(state_season_profit)

best_season_per_state = state_season_profit.idxmax(axis=1)
consistent = (best_season_per_state == best_profit_season).sum()
total_states = len(best_season_per_state)
print(f"\n→ {consistent} out of {total_states} states agree with the overall trend")
print(f"  that {best_profit_season} is the most profitable season.")
```

```python
plt.figure(figsize=(12,6))
sns.boxplot(x='Crop', y='Yield_Tonnes_Ha', hue='Season', data=df)
plt.xticks(rotation=45)
plt.title('Yield by Crop and Season')
plt.tight_layout()
plt.show()
```
